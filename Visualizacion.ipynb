```python
import pandas as pd
import plotly.express as px
df = pd.read_csv('Datos/StudentPerformanceFactors.csv')
```

```python
# Function para actualizar el layout de las graficas
def actualiza_layout(grafica, x_title, y_title):
    grafica.update_layout(
        xaxis_title=x_title,
        yaxis_title=y_title,
        paper_bgcolor='gray',
        plot_bgcolor='gray',
        title_pad_t=20,
        title_font_family='verdana',
        title_font_color='black',
        title_font_size=16,
        font_size=15,
        height=400,
        width=600
    )
```

```python
# Gráfica de dispersión
def sct(varX, varY, co, cocs, x_title, y_title, tam, marg_x=None, marg_y=None, facetCol=None):
    grafica_sc = px.scatter(df, x=varX, y=varY,
                          color=co,
                          color_continuous_scale=cocs,
                          size=tam,
                          marginal_x=marg_x,
                          marginal_y=marg_y,
                          facet_col=facetCol)

    actualiza_layout(grafica_sc, x_title, y_title)
    grafica_sc.show()
```

```python
def histograma(var, tit, subtit, col, cods, textoA, x_titulo, y_titulo,
               agrupados, x_min=0, x_max=0, amplitude=0,
               varY=None, pshape=None):

    if agrupados == True:
        grafica_hist = px.histogram(df, x=var,
                                  title=tit,
                                  subtitle=subtit,
                                  text_auto=textoA)

        # Solo aplicar bins si se especifican los parámetros
        if x_min != 0 or x_max != 0 or amplitude != 0:
            grafica_hist.update_traces(marker_line_width=1,
                                     xbins=dict(start=x_min, end=x_max, size=amplitude))

    elif agrupados == False:
        grafica_hist = px.histogram(df, x=var, y=varY, pattern_shape=pshape,
                                  title=tit,
                                  subtitle=subtit,
                                  color=col,
                                  color_discrete_sequence=cods,
                                  text_auto=textoA)

    actualiza_layout(grafica_hist, x_titulo, y_titulo)
    grafica_hist.show()
```

```python
def intervalos(var):
    import math
    n = len(var)
    x_max = var.max()
    x_min = var.min()
    recorrido = x_max - x_min
    intervalos = round(1 + (3.3 * (math.log10(n))))
    amplitud = (recorrido / intervalos)
    return x_min, x_max, amplitud
```

```python
from plotly.figure_factory import create_distplot

def grafica_densidad(var, etiqueta, color):
    _, _, amplitudA = intervalos(var)
    amplitud = amplitudA

    grafica_densidad = create_distplot([var], [etiqueta],
        show_hist = True,
        show_curve = True,
        curve_type = 'kde',
        show_rug = False,
        bin_size = amplitud,
        colors = [color]
    )

    grafica_densidad.update_traces(marker_line_width = 1)

    grafica_densidad.update_layout(
        title = 'Gráfica de densidad de la variable ' + etiqueta,
        xaxis_title = 'Rango de ' + etiqueta,
        yaxis_title = 'Frecuencia - Densidad',
        width = 600,
        height = 400
    )

    grafica_densidad.show()
```

```python
# Gráfica de densidad para la variable 'Sleep_Hours'
var = df['Sleep_Hours']
etiq = 'Horas de Sueño'
color = 'blue'
grafica_densidad(var, etiq, color)
```

```python
# Gráfica de densidad para la variable 'Previous_Scores'
var = df['Previous_Scores']
etiq = 'Puntajes Previos'
color = 'green'
grafica_densidad(var, etiq, color)
```

```python
def boxplt1(yvar, cds, titulo, x_titulo, y_titulo, col=None, xvar=None, puntos=None):
    grafica_box = px.box(df, x=xvar, y=yvar,
        points=puntos,
        color=col,
        color_discrete_sequence=cds,
        title=titulo
    )
    actualiza_layout(grafica_box, x_titulo, y_titulo)
    grafica_box.show()
```
